```python
#import necessary assets

import warnings
warnings.simplefilter("ignore") 

import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix

from sklearn.metrics import r2_score
```

```python
#read the data from the dataset
cars = pd.read_csv("car_dataset.csv")
```

```python
#visualize the data
cars.head()
```

```python
cars.info()
```

```python
# Use the .isnull() method to locate missing data
missing_values = cars.isnull()
missing_values
```

```python
#drop duplicates
```

```python
#drop unnecessary variables
cars = cars[["selling_price", "km_driven", "mileage", "seats"]].dropna()

#drop rows with missing values
cars.dropna(inplace=True)

cars.head()
```

```python
#convert mileage to numerical date by taking only the digits
cars['mileage'] = cars['mileage'].str.extract(r'(\d+\.?\d*)').astype(float)

#i used google gemini to learn how to extract the numerical values
```

```python
#use scaler
```

```python
# Split data into 'X' features and 'y' target label sets
X = cars[['km_driven', 'mileage', 'seats']]

#the target is y because we want to predict how much a car will cost
y = cars['selling_price']
```

```python
# Split data set into training and test sets
#use a 70% training/30% testing split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=100)
```

```python
#create and train model: Elastic Net Regression
#(selected becuase Elastic Net is good for predicting prices based on inputs)

# Initialize a model
ENModel = ElasticNet(alpha=1.0, l1_ratio=0.5)

#fit the model
ENModel.fit(X_train, y_train)
```

```python
#do a little test

# Predict the price of a car with 150000 miles, 23.5 kmpl, and 5 seats
prediction = ENModel.predict([[150000, 23.5, 5.0]])
print("Predicted price is", np.round(prediction, 2))

#looks right to me
```

```python
#evaluate the model using r-squared
r2 = ENModel.score(X_test, y_test)

print(r2)

#i used google gemini to learn how to use the r-squared function
```
